This part of the pipeline annotates all pangenome protein families using the COG system via eggNOG-mapper. This specific notebook runs `eggNOG-mapper`.

### Checking dependencies

```bash
conda activate eggnog_mapper
emapper.py --version
conda deactivate
```

### Paths and parameters

#### Pipeline input folders

```bash
pangenomes="03-pangenomes"
representative_proteome="$pangenomes/all/all_protein_families.faa"
```

#### Pipeline output folders

```bash
task_root="06-pangenome_annotation"
COG_mapper="$task_root/mapper"

mkdir -p $task_root $COG_mapper
```

#### Tool pointers and parameters

```bash
n_cpus=22
eggnog_db="/mnt/STORAGE/databases/eggnog_mapper"
```

### Running the mapper

```bash
conda activate eggnog_mapper
```

settings mirroring the webserver's defaults

```bash
emapper.py --cpu $n_cpus --mp_start_method forkserver \
--data_dir $eggnog_db -o all --output_dir $COG_mapper --override -m diamond --dmnd_ignore_warnings -i $representative_proteome \
--evalue 0.001 --score 60 --pident 40 --query_cover 20 --subject_cover 20 --itype proteins
```

```bash
conda deactivate
```
